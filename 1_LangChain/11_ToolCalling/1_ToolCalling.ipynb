```python
import warnings
warnings.filterwarnings("ignore")
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

```python
# Creating tool
@tool 
def multiply(a:int , b:int) -> int:
    """Given 2 numbers a and b and this tool return their product"""
    return a*b

# tool binding with llm
llm = ChatGoogleGenerativeAI(
    model = "gemini-3.6-flash"
)
llm_with_tool = llm.bind_tools([multiply])

# query
messages = list()

    # human message
human_message = HumanMessage(
    content = "can you multiply 3 and 6"
)
messages.append(human_message)

    # AI message
ai_message = llm_with_tool.invoke(messages)
messages.append(ai_message)

    # tool message
tool_message = multiply.invoke(
    ai_message.tool_calls[0]
)
messages.append(tool_message)

final_res = llm_with_tool.invoke(messages)
print(final_res)
```
