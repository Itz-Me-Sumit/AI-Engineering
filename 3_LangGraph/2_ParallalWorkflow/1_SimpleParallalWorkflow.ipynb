```python
from langgraph.graph import StateGraph , START , END
from typing import TypedDict
```

```python
class cricket(TypedDict):
    runs:int
    balls:int
    fours:int
    sixs:int
    strike_rate:float
    runs_in_boundary:float
    balls_per_boundary:float
```

```python
"""
def strike_rate(state:cricket) -> cricket:
    state["strike_rate"] = (state["runs"]/state["balls"]) * 100
    return state

def runs_in_boundary(state:cricket) -> cricket:
    state["runs_in_boundary"] = (4 * state["fours"]) + (6 * state["sixs"])
    return state

def balls_per_boundary(state:cricket) -> cricket:
    state["balls_per_boundary"] = state["balls"] / (state["fours"] + state["sixs"])
    return state
"""

def strike_rate(state: cricket):
    return {
        "strike_rate": (state["runs"] / state["balls"]) * 100
    }


def runs_in_boundary(state: cricket):
    return {
        "runs_in_boundary": (4 * state["fours"]) + (6 * state["sixs"])
    }


def balls_per_boundary(state: cricket):
    return {
        "balls_per_boundary": state["balls"] / (state["fours"] + state["sixs"])
    }
```

```python
graph = StateGraph(cricket)

# Node
graph.add_node("strike_rate" , strike_rate)
graph.add_node("runs_in_boundary" , runs_in_boundary)
graph.add_node("balls_per_boundary" , balls_per_boundary)

# Edges
graph.add_edge(START , "strike_rate")
graph.add_edge("strike_rate" , END)

graph.add_edge(START , "runs_in_boundary")
graph.add_edge("runs_in_boundary" , END)

graph.add_edge(START , "balls_per_boundary")
graph.add_edge("balls_per_boundary" , END)

# Compile
workflow = graph.compile()
```

```python
initial_state = {
    "runs" : 72,
    "balls" : 48,
    "fours" : 6,
    "sixs" : 4,
}

final_state = workflow.invoke(initial_state)
```
