```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

client = OpenAI(api_key = os.getenv("OPENAI_API_KEY"))

prompt = "Explain Transformers with a begineer fraindly short explaination"

response = client.chat.completions.create(
    model = "gpt-4o-mini",
    messages = [
        {
            "role" : "user",
            "content" : prompt
        }
    ],
    max_tokens = 250
)

print(f"User : {prompt}\n\nModel Response : {response.choices[0].message.content.strip()}")
```
