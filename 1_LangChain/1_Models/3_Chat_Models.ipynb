#### Open AI

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

load_dotenv()

OpenAI_chat_model = ChatOpenAI(
    model = "gpt-4.1-mini",
    temperature = 0,
    max_completion_tokens = 100
)

result = OpenAI_chat_model.invoke("What is the capital of india")

print(result)
```

#### Claude

```python
from langchain_anthropic import ChatAnthropic
from dotenv import load_dotenv

load_dotenv()

Claude_chat_model = ChatAnthropic(
    model = "claude-3-5-sonnet-20241022",
    temperature = 0,
    max_tokens = 100
)

result = Claude_chat_model.invoke("Hello Can you have some mercy and give me lil free tokens ")

print(result)
```

#### Gemini

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv

load_dotenv()

Gemini_chat_model = ChatGoogleGenerativeAI(
    model = "gemini-1.5-pro",
    temperature = 0,
    max_tokens = 100
)

result = Gemini_chat_model.invoke("Hey What's up my bro give me some free Tokens Man")

print(result)
```

#### Hugging Face (API)

```python
from langchain_huggingface import ChatHuggingFace , HuggingFaceEndpoint
from dotenv import load_dotenv

load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id = "meta-llama/Llama-3.1-8B-Instruct",
    task = "text-generation",
    max_new_tokens=100
)

HuggingFace_chat_model = ChatHuggingFace(llm = llm)

result = HuggingFace_chat_model.invoke("what is the capital of india")

print(result.content)
```

#### Hugging Face (Local)

```python
from langchain_huggingface import ChatHuggingFace , HuggingFacePipeline
from dotenv import load_dotenv

load_dotenv()

llm = HuggingFacePipeline.from_model_id(
    model_id = "HuggingFaceTB/SmolLM2-1.7B-Instruct",
    task = "text-generation",
    pipeline_kwargs = {
        "temperature" : 0.5,
        "max_new_tokens" : 100,
    }
    
)

model = ChatHuggingFace(llm = llm)

model.invoke("What is the capital of india")
```
