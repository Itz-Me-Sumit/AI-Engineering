```python
import warnings
warnings.filterwarnings("ignore")

from langgraph.graph import StateGraph , START , END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(
    model = "gemini-3.6-flash"
)
```

```python
# Creating state
class LLMState(TypedDict):
    question:str
    answer:str
```

```python
def llm_qa(state:LLMState) -> LLMState:

    query = state["question"]

    prompt = f"Give the response of following query in 50 words\n{query}"

    response = model.invoke(prompt)

    state["answer"] = response.content[0]["text"]

    return state
```

```python
# creating graph
graph = StateGraph(LLMState)

# Creating Node
graph.add_node("llm_response" , llm_qa)

# Creating Edge
graph.add_edge(START , "llm_response")
graph.add_edge("llm_response" , END)

# compile graph
workflow = graph.compile()

# execute graph
```

```python
initial_state = {
    "question" : "tell me something excited about AI",
    "answer" : ""
}

final_state = workflow.invoke(initial_state)

print(final_state)
```
